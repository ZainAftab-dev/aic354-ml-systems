## Step 1: Import Libraries

> **Understanding:** All tools are loaded first: **Pandas** for tables, **NumPy** for arrays, **LabelEncoder** to turn text into numbers, **SVC** (Support Vector Classifier) as the learning algorithm, **pickle** to save the model, and metrics to measure how good the model is.

````python
"""
Purpose of this code:
----------------------
This script imports the required libraries for building and evaluating a
machine learning model. It sets up tools for data manipulation, preprocessing,
splitting datasets, training a Support Vector Machine (SVM) model,
evaluating accuracy, and displaying results in tabular form.
"""

# Import numerical computing library (used for arrays, math operations, etc.)
import numpy as np

# Import data manipulation library (used for DataFrames, reading/writing data, etc.)
import pandas as pd

# Import pickle (used for saving and loading Python objects/models)
import pickle

# Import train_test_split (used to divide dataset into training and testing sets)
from sklearn.model_selection import train_test_split

# Import LabelEncoder (used to convert categorical values into numeric codes)
from sklearn.preprocessing import LabelEncoder

# Import Support Vector Machine (SVM) model from scikit-learn
from sklearn import svm

# Import evaluation metrics
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Import PrettyTable (used to display results in a clean, table format)
from prettytable import PrettyTable

# Import plotting libraries (used for the confusion matrix heatmap)
import matplotlib.pyplot as plt
import seaborn as sns
````

## Step 2: Load Sample Data

> **Understanding:** The 100-row sample dataset is read from a CSV file into a DataFrame. Every row is one passenger; the last column (`Survived`) is the output we want to predict.

````python
"""
Purpose of this section:
-------------------------
This part of the code loads a sample dataset from a CSV file into a
pandas DataFrame. It then displays the dataset so that students can
see the raw data being used for further processing and model training.
"""

# Read the CSV file and load it into a pandas DataFrame
# 'titanic/sample-data.csv' should be present in the working directory
sample_data = pd.read_csv("titanic/sample-data.csv")

# Print a header to indicate that sample data is being displayed
print("\n\nSample Data:")
print("============\n")

# Configure pandas to display all rows and columns (so nothing gets hidden with '...')
pd.set_option("display.max_rows", None, "display.max_columns", None)

# Print the complete dataset
print(sample_data)
````

## Step 3: Understand and Pre-process Sample Data

### Step 3.1: Understand Sample Data

> **Understanding:** Before building anything we check *what* the data contains: 5 attributes (4 inputs + 1 output) and 100 instances.

````python
"""
Purpose of this section:
-------------------------
This part of the code helps us understand the dataset by:
1. Printing the names of all attributes (columns) in the DataFrame.
2. Showing the total number of instances (rows) available in the dataset.
"""

# Print a header for clarity
print("\n\nAttributes in Sample Data:")
print("==========================\n")

# Display all column names (attributes) in the dataset
print(sample_data.columns)

# Print a header for instance count
print("\n\nNumber of Instances in Sample Data:", sample_data["PClass"].count())
print("========================================\n")
````

### Step 3.2: Pre-process Sample Data
- Sample Data is already Preprocessed
- No Preprocessing needs to be Performed

> **Understanding:** Pre-processing means cleaning the data (missing values, wrong types, inconsistent text). This sample was prepared in advance, so no cleaning is needed here. In Part B real cleaning is required.

## Step 4: Feature Extraction
- Features are already Extracted
- No Feature Extraction needs to be Performed

> **Understanding:** Feature extraction means choosing or creating the attributes the model will learn from. The four input attributes are already selected.

## Step 5: Label Encoding the Sample Data (Input and Output is converted in Numeric Representation)

### Step 5.1: Train the Label Encoder

> **Understanding:** ML algorithms only understand numbers. A **LabelEncoder** is trained on the list of possible values of each attribute and gives each value a number in **alphabetical order** (for example `Cherbourg = 0, Queenstown = 1, Southampton = 2`).

````python
"""
Purpose of this section:
-------------------------
This part of the code sets up and trains Label Encoders.
Label Encoding is used to convert categorical values (like 'Male', 'Female', 'First', 'Second')
into numeric codes that machine learning algorithms can work with.

Steps:
1. Define sample label categories for each attribute.
2. Initialize a LabelEncoder for each attribute.
3. Train (fit) each LabelEncoder with its respective category values.
"""

# Define label categories for each feature
# These are possible values each column can have in the dataset
pclass = pd.DataFrame({"Pclass": ["First", "Second", "Third"]})
gender = pd.DataFrame({"Gender": ["Male", "Female"]})
sibling = pd.DataFrame({"Sibling": ["Zero", "One", "Two", "Three"]})
embarked = pd.DataFrame({"Embarked": ["Southampton", "Cherbourg", "Queenstown"]})
survived = pd.DataFrame({"Survived": ["Yes", "No"]})

# Create LabelEncoder objects for each categorical column
pclass_label_encoder = LabelEncoder()
gender_label_encoder = LabelEncoder()
sibling_label_encoder = LabelEncoder()
embarked_label_encoder = LabelEncoder()
survived_label_encoder = LabelEncoder()

# Train (fit) the encoders on the given categories
# np.ravel() flattens the DataFrame column into a 1D array for the encoder
pclass_label_encoder.fit(np.ravel(pclass))
gender_label_encoder.fit(np.ravel(gender))
sibling_label_encoder.fit(np.ravel(sibling))
embarked_label_encoder.fit(np.ravel(embarked))
survived_label_encoder.fit(np.ravel(survived))
````

### Step 5.2: Label Encode the Output

> **Understanding:** The output `Survived` is converted: `No = 0`, `Yes = 1`.

````python
"""
Purpose of this section:
-------------------------
This part of the code applies Label Encoding to the target/output variable
("Survived") in the dataset. The categorical values ("Yes", "No") are
converted into numerical values (e.g., 1 and 0) that can be used by
machine learning algorithms.
"""

# Create copies of the dataset (one for encoded output, one to keep original)
sample_data_encoded_output = sample_data.copy()
original_sample_data = sample_data.copy()

# Transform "Survived" (categorical -> numeric) using the trained LabelEncoder
print("\n\nSurvived Attribute After Label Encoding:")
print("========================================\n")
sample_data["encoded_survived"] = survived_label_encoder.transform(sample_data['Survived'])

# Show original "Survived" values and their numeric encodings
print(sample_data[["Survived", "encoded_survived"]])

# Replace original categorical columns with encoded values for the output dataset
sample_data_encoded_output[['PClass', 'Gender', 'Sibling', 'Embarked', 'Survived']] = \
    sample_data[['PClass', 'Gender', 'Sibling', 'Embarked', 'encoded_survived']]

# Save the encoded dataset to a CSV file (useful for training/testing later)
sample_data_encoded_output.to_csv(r'titanic/sample-data-encoded-output.csv', index=False, header=True)
print("\nSaved: titanic/sample-data-encoded-output.csv")
````

### Step 5.3: Label Encode the Input

> **Understanding:** The four input attributes are converted with their own encoders. The fully numeric table is saved as `sample-data-encoded.csv`.

````python
"""
Purpose of this section:
-------------------------
This part of the code applies Label Encoding to the input features
("PClass", "Gender", "Sibling", "Embarked"). Each categorical value
is converted into a numeric value so that machine learning models
can process them.
"""

# Create copies of the dataset (to keep original values safe)
sample_data_encoded = sample_data_encoded_output.copy()

# Encode each input attribute
sample_data_encoded_output["encoded_pclass"] = pclass_label_encoder.transform(sample_data_encoded_output['PClass'])
sample_data_encoded_output["encoded_gender"] = gender_label_encoder.transform(sample_data_encoded_output['Gender'])
sample_data_encoded_output["encoded_sibling"] = sibling_label_encoder.transform(sample_data_encoded_output['Sibling'])
sample_data_encoded_output["encoded_embarked"] = embarked_label_encoder.transform(sample_data_encoded_output['Embarked'])

# Replace categorical columns with encoded values in the dataset
sample_data_encoded[['PClass', 'Gender', 'Sibling', 'Embarked', 'Survived']] = \
    sample_data_encoded_output[['encoded_pclass', 'encoded_gender', 'encoded_sibling', 'encoded_embarked', 'Survived']]

# Print encoded dataset (with numeric values instead of categories)
print("\n\nSample Data after Label Encoding:")
print("=================================\n")
print(sample_data_encoded)

# Save the encoded dataset to a new CSV file
sample_data_encoded.to_csv(r'titanic/sample-data-encoded.csv', index=False, header=True)
````

## Step 6: Execute the Training Phase

### Step 6.1: Splitting Sample Data into Training Data and Testing Data

> **Understanding:** **Training Phase begins.** The data is split into **80% training data** (80 passengers, used to learn) and **20% testing data** (20 passengers, kept hidden to check the model later). `shuffle=False` keeps the original row order, so the split is the same every time.

````python
"""
Purpose of this section:
-------------------------
This part of the code splits the dataset into Training Data and Testing Data.
- Training Data is used to teach the machine learning model.
- Testing Data is used to check how well the model performs on unseen data.
"""

# Split dataset into training (80%) and testing (20%)
# random_state=0 ensures reproducibility (same split every time)
# shuffle=False keeps the original order of rows (no shuffling before splitting)
training_data_encoded, testing_data_encoded = train_test_split(
    sample_data_encoded,
    test_size=0.2,
    random_state=0,
    shuffle=False
)

# Save the training and testing datasets to CSV files
training_data_encoded.to_csv(r'titanic/training-data-encoded.csv', index=False, header=True)
testing_data_encoded.to_csv(r'titanic/testing-data-encoded.csv', index=False, header=True)

print("Training Data shape:", training_data_encoded.shape)
print("Testing Data shape:", testing_data_encoded.shape)
````

### Step 6.2: Splitting Input Vectors and Outputs / Labels of Training Data

> **Understanding:** The training table is divided into **input vectors** (first 4 columns, `X`) and **output labels** (last column, `y`).

````python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (input vectors) and
the labels (output values) from the training dataset.
"""

# Extract input vectors (features) from training data
input_vector_train = training_data_encoded.iloc[:, :-1]

# Extract output labels (target variable) from training data
output_label_train = training_data_encoded.iloc[:, -1]

print("Input vectors (train) shape:", input_vector_train.shape)
print("Output labels (train) shape:", output_label_train.shape)
````

### Step 6.3: Train the Support Vector Classifier

> **Understanding:** The **Support Vector Classifier** finds the boundary that best separates survivors from non-survivors. `fit()` is the actual learning step.

````python
"""
Purpose of this section:
-------------------------
This part of the code trains a Support Vector Classifier (SVC) on the training data.
The SVC is a type of machine learning model that finds the best boundary
to separate different classes (e.g., survived vs. not survived).
"""

print("\n\nTraining the Support Vector Classifier on Training Data")
print("========================================================\n")

# Initialize the Support Vector Classifier with chosen parameters
svc_model = svm.SVC(gamma='auto', random_state=0)

# Train the model using the training data (features and labels)
svc_model.fit(input_vector_train, np.ravel(output_label_train))

# Print trained model details (parameters and values)
print(svc_model)
````

### Step 6.4: Save the Trained Model

> **Understanding:** The trained model is saved to a `.pkl` file, so it can be reused (testing, application, deployment) without retraining.

````python
"""
Purpose of this section:
-------------------------
This part of the code saves the trained Support Vector Classifier (SVC)
model to your computer so it can be reused later without retraining.
"""

# Save the trained SVC model to a file named "svc_trained_model.pkl"
pickle.dump(svc_model, open('titanic/svc_trained_model.pkl', 'wb'))
print("Saved: titanic/svc_trained_model.pkl")
````

## Step 7: Execute the Testing Phase

### Step 7.1: Splitting Input Vectors and Outputs/Labels of Testing Data

> **Understanding:** **Testing Phase begins.** The 20 testing passengers are split into inputs and true labels, exactly like the training data.

````python
"""
Purpose of this section:
-------------------------
This part of the code separates the features (inputs) and the labels (outputs)
from the testing dataset.
"""

input_vector_test = testing_data_encoded.iloc[:, :-1]
output_label_test = testing_data_encoded.iloc[:, -1]

print("Input vectors (test) shape:", input_vector_test.shape)
print("Output labels (test) shape:", output_label_test.shape)
````

### Step 7.2: Load the Saved Model

> **Understanding:** The saved model is loaded back from disk, which proves the saved file works.

````python
"""
Purpose of this section:
-------------------------
This part of the code loads the previously saved Support Vector Classifier (SVC)
model from disk into memory.
"""

model = pickle.load(open('titanic/svc_trained_model.pkl', 'rb'))
print("Model loaded from titanic/svc_trained_model.pkl")
````

### Step 7.3: Evaluate the Machine Learning Model

#### Step 7.3.1: Make Predictions with the Trained Models on Testing Data

> **Understanding:** The model predicts the 20 unseen passengers. The `Predictions` column is placed next to the true `Survived` column for comparison.

````python
"""
Purpose of this section:
-------------------------
This part of the code evaluates the trained Support Vector Classifier (SVC)
on the testing data.
"""

model_predictions = model.predict(input_vector_test)

pd.options.mode.chained_assignment = None
testing_data_encoded["Predictions"] = model_predictions
testing_data_encoded.to_csv(r'titanic/model-predictions.csv', index=False, header=True)

model_predictions = testing_data_encoded
print("\n\nPredictions Returned by svc_trained_model:")
print("==========================================\n")
print(model_predictions)
````

### Step 7.4: Calculate the Accuracy Score

> **Understanding:** **Accuracy** = correct predictions / total predictions. An accuracy of 0.80 means 16 of the 20 testing passengers were predicted correctly.

````python
"""
Purpose of this section:
-------------------------
This part of the code calculates how accurate the trained model is on the
testing dataset.
"""

model_accuracy_score = accuracy_score(
    model_predictions["Survived"],
    model_predictions["Predictions"]
)

print("\n\nAccuracy Score:")
print("===============\n")
print(round(model_accuracy_score, 2))
````

### Step 7.5: Detailed Evaluation Metrics (Precision, Recall, F1, Confusion Matrix)

> **Understanding:** Accuracy alone can hide errors. **Precision**, **Recall**, **F1-score** and the **Confusion Matrix** show exactly which class the model gets wrong.

````python
"""
Purpose of this section:
-------------------------
Accuracy alone can be misleading, especially on small or imbalanced
datasets. This section adds the standard classification metrics:
- Precision: of everyone the model predicted as "Survived", how many actually survived?
- Recall: of everyone who actually survived, how many did the model catch?
- F1-score: the harmonic mean of precision and recall.
- Confusion Matrix: a table of correct vs. incorrect predictions per class.
"""

print("\nClassification Report:")
print("=======================\n")
print(classification_report(
    model_predictions["Survived"],
    model_predictions["Predictions"],
    target_names=["Not Survived (0)", "Survived (1)"]
))

conf_matrix = confusion_matrix(model_predictions["Survived"], model_predictions["Predictions"])
conf_matrix_df = pd.DataFrame(
    conf_matrix,
    index=["Actual: Not Survived", "Actual: Survived"],
    columns=["Predicted: Not Survived", "Predicted: Survived"]
)
print("\nConfusion Matrix:")
print("==================\n")
print(conf_matrix_df)

plt.figure(figsize=(5, 4))
sns.heatmap(conf_matrix, annot=True, fmt='d', cmap='Blues',
            xticklabels=['Not Survived', 'Survived'],
            yticklabels=['Not Survived', 'Survived'])
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix - SVC on Titanic Testing Data')
plt.tight_layout()
plt.show()
````

## Step 8: Execute the Application Phase

### Step 8.1: Take Input from User

> **Understanding:** **Application Phase begins.** A new passenger that the model has never seen is entered (pre-filled here so the notebook runs from top to bottom).

````python
"""
Purpose of this section:
-------------------------
This part of the code allows the user to enter their own data
(PClass, Gender, Sibling, Embarked). In an interactive Jupyter session,
running this cell will prompt you for input(). For this notebook copy,
we pre-fill a demo passenger so the notebook can also be read/run
top-to-bottom without needing to type anything.

To try your own values interactively, replace the four lines below with:
    pclass_input = input("Please enter PClass here (First, Second, Third): ").strip()
    gender_input = input("Please enter your Gender here (Male, Female): ").strip()
    sibling_input = input("Please enter your Sibling here (Zero, One, Two, Three): ").strip()
    embarked_input = input("Please enter Embarked here (Cherbourg, Southampton, Queenstown): ").strip()
"""

# Demo passenger (edit these values, or swap in input() calls as shown above)
pclass_input = "First"
gender_input = "Male"
sibling_input = "Zero"
embarked_input = "Queenstown"

print(f"PClass: {pclass_input}, Gender: {gender_input}, Sibling: {sibling_input}, Embarked: {embarked_input}")
````

### Step 8.2: Convert User Input into Feature Vector (Exactly Same as Feature Vectors of Sample Data)

> **Understanding:** The raw input is arranged into a one-row DataFrame with the **same column names and order** as the training data.

````python
"""
Purpose of this section:
-------------------------
This part of the code takes the user's input (entered as text)
and converts it into a structured feature vector (DataFrame).
"""

user_input = pd.DataFrame({
    'PClass': [pclass_input],
    'Gender': [gender_input],
    'Sibling': [sibling_input],
    'Embarked': [embarked_input]
})

print("\n\nUser Input Feature Vector:")
print("==========================\n")
print(user_input)
````

### Step 8.3: Label Encoding of Feature Vector (Exactly Same as Label Encoded Feature Vectors of Sample Data)

> **Understanding:** The input is encoded with the **same trained encoders**. Using different encoders would give the model wrong numbers.

````python
"""
Purpose of this section:
-------------------------
This part of the code encodes the user's input into numeric values
using the previously trained LabelEncoders.
"""

unseen_data_features = user_input.copy()

unseen_data_features["PClass"] = pclass_label_encoder.transform(user_input['PClass'])
unseen_data_features["Gender"] = gender_label_encoder.transform(user_input['Gender'])
unseen_data_features["Sibling"] = sibling_label_encoder.transform(user_input['Sibling'])
unseen_data_features["Embarked"] = embarked_label_encoder.transform(user_input['Embarked'])

print("\n\nUser Input Encoded Feature Vector:")
print("==================================\n")
print(unseen_data_features)
````

### Step 8.4: Load the Saved Model

### Problem Definition
A **binary classification** task: predict whether a patient **has heart disease (Yes)** or **does not (No)** from four clinical attributes.

### Dataset
| | |
|---|---|
| **Source** | UCI Machine Learning Repository, *Heart Disease* dataset (ID 45): https://archive.ics.uci.edu/dataset/45/heart+disease |

## Step 1: Import Libraries

> **Understanding:** The same libraries as the Titanic system are used, so the two systems can be compared step by step.

````python
"""
Purpose of this code:
----------------------
Imports the same libraries used in the Titanic system: data handling,
label encoding, data splitting, the Support Vector Classifier, evaluation
metrics, PrettyTable for output, and plotting libraries.
"""

import numpy as np                      # numerical arrays
import pandas as pd                     # DataFrames (tables)
import pickle                           # save / load the trained model
from sklearn.model_selection import train_test_split   # split training / testing data
from sklearn.preprocessing import LabelEncoder          # text -> numbers
from sklearn import svm                                 # Support Vector Machine
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
from prettytable import PrettyTable                     # neat output table
import matplotlib.pyplot as plt                         # plots
import seaborn as sns                                   # confusion-matrix heatmap
````

## Step 2: Load Sample Data

> **Understanding:** The original UCI file has **no header row**, so the 14 column names are given from the dataset's documentation (`heart-disease.names`). Missing values are written as `?` in the file, so `na_values="?"` tells Pandas to read them as missing (`NaN`).

````python
"""
Purpose of this section:
-------------------------
Loads the original Cleveland Heart Disease data downloaded from the UCI
Machine Learning Repository into a pandas DataFrame.
"""

# Column names from the UCI documentation (the file itself has no header)
column_names = ["age", "sex", "cp", "trestbps", "chol", "fbs", "restecg",
                "thalach", "exang", "oldpeak", "slope", "ca", "thal", "num"]

# Read the raw file; "?" means a missing value
raw_data = pd.read_csv("heart_disease/processed.cleveland.data", names=column_names, na_values="?")

print("\n\nRaw Heart Disease Data (first 10 rows):")
print("=======================================\n")
pd.set_option("display.max_columns", None)
print(raw_data.head(10))
````

## Step 3: Understand and Pre-process Sample Data

### Step 3.1: Understand Sample Data

> **Understanding:** Unlike the Titanic sample, this is real data, so we look at more than the column names: the number of instances, the missing values and the distribution of the output.

| Attribute | Meaning | Attribute | Meaning |
|---|---|---|---|
| `age` | Age in years | `thalach` | Maximum heart rate achieved |
| `sex` | 1 = male, 0 = female | `exang` | Exercise-induced angina (1 = yes, 0 = no) |
| `cp` | Chest pain type (1 to 4) | `oldpeak` | ST depression caused by exercise |
| `trestbps` | Resting blood pressure (mm Hg) | `slope` | Slope of the peak exercise ST segment |
| `chol` | Serum cholesterol (mg/dl) | `ca` | Major vessels coloured by fluoroscopy (0 to 3) |
| `fbs` | Fasting blood sugar > 120 mg/dl | `thal` | 3 = normal, 6 = fixed defect, 7 = reversible defect |
| `restecg` | Resting ECG result (0 to 2) | `num` | **Output:** 0 = no disease, 1 to 4 = disease present |

````python
"""
Purpose of this section:
-------------------------
Helps us understand the dataset by printing:
1. The names of all attributes.
2. The number of instances.
3. The number of missing values in each attribute.
4. The distribution of the output attribute (num).
"""

print("\n\nAttributes in Sample Data:")
print("==========================\n")
print(raw_data.columns)

print("\n\nNumber of Instances in Sample Data:", raw_data["age"].count())
print("========================================\n")

print("Missing Values in Each Attribute:")
print("=================================\n")
print(raw_data.isnull().sum())

print("\n\nDistribution of the Output Attribute (num):")
print("=============================================\n")
print(raw_data["num"].value_counts().sort_index())
````

### Step 3.2: Pre-process Sample Data

> **Understanding:** Two pre-processing steps are needed (the Titanic sample needed none):
> 1. **Remove missing values:** 6 rows contain a `?`. They are only 2% of the data, so they are deleted.
> 2. **Convert the output into two classes:** `num` has 5 values (0 to 4). Following the UCI documentation, 0 means *no heart disease* and 1, 2, 3, 4 all mean *heart disease present*. So `num` becomes `HeartDisease` with the values **No** and **Yes**, just like `Survived` in the Titanic data.

````python
"""
Purpose of this section:
-------------------------
1. Removes the rows that contain missing values.
2. Converts the output attribute "num" (0 to 4) into "HeartDisease" (No / Yes).
"""

# 1. Remove rows with missing values
clean_data = raw_data.dropna().copy()
print("Instances before removing missing values:", len(raw_data))
print("Instances after removing missing values :", len(clean_data))

# 2. Output: 0 -> "No", 1 to 4 -> "Yes"
clean_data["HeartDisease"] = clean_data["num"].apply(lambda value: "No" if value == 0 else "Yes")

print("\n\nHeartDisease Attribute after Conversion:")
print("========================================\n")
print(clean_data["HeartDisease"].value_counts())
````

## Step 4: Feature Extraction

> **Understanding:** This is where the assignment's constraint is applied: only **4 input attributes** are selected out of 13. Their numeric codes are converted into readable text values (for example `cp = 4` becomes `Asymptomatic`), so the sample data looks exactly like the Titanic sample data. The result is saved as `heart_disease/sample-data.csv`.

````python
"""
Purpose of this section:
-------------------------
1. Selects the 4 input attributes (cp, slope, ca, thal) and the output.
2. Converts their numeric codes into readable categories (as in the Titanic data).
3. Saves the sample data to a CSV file.
"""

# Code -> readable value, taken from the UCI documentation
chest_pain_names = {1: "Typical Angina", 2: "Atypical Angina", 3: "Non-Anginal Pain", 4: "Asymptomatic"}
slope_names = {1: "Upsloping", 2: "Flat", 3: "Downsloping"}
major_vessels_names = {0: "Zero", 1: "One", 2: "Two", 3: "Three"}
thal_names = {3: "Normal", 6: "Fixed Defect", 7: "Reversible Defect"}

sample_data = pd.DataFrame({
    "ChestPain": clean_data["cp"].astype(int).map(chest_pain_names),
    "Slope": clean_data["slope"].astype(int).map(slope_names),
    "MajorVessels": clean_data["ca"].astype(int).map(major_vessels_names),
    "Thal": clean_data["thal"].astype(int).map(thal_names),
    "HeartDisease": clean_data["HeartDisease"],
}).reset_index(drop=True)

# Save the sample data (same role as titanic/sample-data.csv)
sample_data.to_csv(r"heart_disease/sample-data.csv", index=False, header=True)

print("\n\nSample Data (first 15 of", len(sample_data), "instances):")
print("===============================================\n")
print(sample_data.head(15))
````

## Step 5: Label Encoding the Sample Data (Input and Output is converted in Numeric Representation)

### Step 5.1: Train the Label Encoder

> **Understanding:** One LabelEncoder is trained per attribute on its list of possible values. The table printed below shows the number given to every value (alphabetical order), which is needed to read the encoded data later.

````python
"""
Purpose of this section:
-------------------------
Sets up and trains one Label Encoder for each attribute.
"""

# Possible values of each attribute
chest_pain = pd.DataFrame({"ChestPain": ["Typical Angina", "Atypical Angina", "Non-Anginal Pain", "Asymptomatic"]})
slope = pd.DataFrame({"Slope": ["Upsloping", "Flat", "Downsloping"]})
major_vessels = pd.DataFrame({"MajorVessels": ["Zero", "One", "Two", "Three"]})
thal = pd.DataFrame({"Thal": ["Normal", "Fixed Defect", "Reversible Defect"]})
heart_disease = pd.DataFrame({"HeartDisease": ["Yes", "No"]})

# Create one LabelEncoder per attribute
chest_pain_label_encoder = LabelEncoder()
slope_label_encoder = LabelEncoder()
major_vessels_label_encoder = LabelEncoder()
thal_label_encoder = LabelEncoder()
heart_disease_label_encoder = LabelEncoder()

# Train (fit) each encoder on its possible values
chest_pain_label_encoder.fit(np.ravel(chest_pain))
slope_label_encoder.fit(np.ravel(slope))
major_vessels_label_encoder.fit(np.ravel(major_vessels))
thal_label_encoder.fit(np.ravel(thal))
heart_disease_label_encoder.fit(np.ravel(heart_disease))

# Show the number given to each value
encoders = {"ChestPain": chest_pain_label_encoder, "Slope": slope_label_encoder,
            "MajorVessels": major_vessels_label_encoder, "Thal": thal_label_encoder,
            "HeartDisease": heart_disease_label_encoder}
codes_table = PrettyTable(["Attribute", "Value", "Code"])
for name, encoder in encoders.items():
    for code_number, value in enumerate(encoder.classes_):
        codes_table.add_row([name, value, code_number])
print(codes_table)
````

### Step 5.2: Label Encode the Output

> **Understanding:** `HeartDisease` is converted: `No = 0`, `Yes = 1`.

````python
"""
Purpose of this section:
-------------------------
Applies Label Encoding to the output attribute "HeartDisease".
"""

sample_data_encoded_output = sample_data.copy()

print("\n\nHeartDisease Attribute After Label Encoding (first 10 rows):")
print("=============================================================\n")
sample_data["encoded_heart_disease"] = heart_disease_label_encoder.transform(sample_data["HeartDisease"])
print(sample_data[["HeartDisease", "encoded_heart_disease"]].head(10))

# Replace the output column with its encoded values
sample_data_encoded_output["HeartDisease"] = sample_data["encoded_heart_disease"]

sample_data_encoded_output.to_csv(r"heart_disease/sample-data-encoded-output.csv", index=False, header=True)
print("\nSaved: heart_disease/sample-data-encoded-output.csv")
````

### Step 5.3: Label Encode the Input

> **Understanding:** The four input attributes are converted with their own encoders. After this step the whole table is numeric and ready for the algorithm.

````python
"""
Purpose of this section:
-------------------------
Applies Label Encoding to the 4 input attributes.
"""

sample_data_encoded = sample_data_encoded_output.copy()

sample_data_encoded["ChestPain"] = chest_pain_label_encoder.transform(sample_data_encoded_output["ChestPain"])
sample_data_encoded["Slope"] = slope_label_encoder.transform(sample_data_encoded_output["Slope"])
sample_data_encoded["MajorVessels"] = major_vessels_label_encoder.transform(sample_data_encoded_output["MajorVessels"])
sample_data_encoded["Thal"] = thal_label_encoder.transform(sample_data_encoded_output["Thal"])

print("\n\nSample Data after Label Encoding (first 15 rows):")
print("=================================================\n")
print(sample_data_encoded.head(15))

sample_data_encoded.to_csv(r"heart_disease/sample-data-encoded.csv", index=False, header=True)
print("\nSaved: heart_disease/sample-data-encoded.csv")
````

## Step 6: Execute the Training Phase

### Step 6.1: Splitting Sample Data into Training Data and Testing Data

> **Understanding:** Same settings as the Titanic system: **80% training** (237 patients) and **20% testing** (60 patients), `random_state=0`, `shuffle=False`.

````python
"""
Purpose of this section:
-------------------------
Splits the dataset into Training Data (80%) and Testing Data (20%).
"""

training_data_encoded, testing_data_encoded = train_test_split(
    sample_data_encoded,
    test_size=0.2,
    random_state=0,
    shuffle=False
)

training_data_encoded.to_csv(r"heart_disease/training-data-encoded.csv", index=False, header=True)
testing_data_encoded.to_csv(r"heart_disease/testing-data-encoded.csv", index=False, header=True)

print("Training Data shape:", training_data_encoded.shape)
print("Testing Data shape:", testing_data_encoded.shape)
````

### Step 6.2: Splitting Input Vectors and Outputs / Labels of Training Data

> **Understanding:** First 4 columns = input vectors, last column = output label.

````python
"""
Purpose of this section:
-------------------------
Separates the input vectors and the output labels of the training data.
"""

input_vector_train = training_data_encoded.iloc[:, :-1]
output_label_train = training_data_encoded.iloc[:, -1]

print("Input vectors (train) shape:", input_vector_train.shape)
print("Output labels (train) shape:", output_label_train.shape)
````

### Step 6.3: Train the Support Vector Classifier

> **Understanding:** The same algorithm with the same settings (`gamma='auto'`, `random_state=0`) learns the boundary between patients with and without heart disease.

````python
"""
Purpose of this section:
-------------------------
Trains a Support Vector Classifier (SVC) on the training data.
"""

print("\n\nTraining the Support Vector Classifier on Training Data")
print("========================================================\n")

svc_model = svm.SVC(gamma="auto", random_state=0)
svc_model.fit(input_vector_train, np.ravel(output_label_train))
print(svc_model)
````

### Step 6.4: Save the Trained Model

> **Understanding:** The trained model is saved as `heart_disease/svc_trained_model.pkl`. This file is uploaded with the submission and used by the cloud app.

````python
"""
Purpose of this section:
-------------------------
Saves the trained SVC model to a file so it can be reused without retraining.
"""

pickle.dump(svc_model, open("heart_disease/svc_trained_model.pkl", "wb"))
print("Saved: heart_disease/svc_trained_model.pkl")
````

## Step 7: Execute the Testing Phase

### Step 7.1: Splitting Input Vectors and Outputs/Labels of Testing Data

> **Understanding:** The 60 testing patients are split into inputs and true labels.

````python
"""
Purpose of this section:
-------------------------
Separates the input vectors and the output labels of the testing data.
"""

input_vector_test = testing_data_encoded.iloc[:, :-1]
output_label_test = testing_data_encoded.iloc[:, -1]

print("Input vectors (test) shape:", input_vector_test.shape)
print("Output labels (test) shape:", output_label_test.shape)
````

### Step 7.2: Load the Saved Model

> **Understanding:** The model is loaded from the `.pkl` file to prove the saved file works.

````python
"""
Purpose of this section:
-------------------------
Loads the saved SVC model from disk.
"""

model = pickle.load(open("heart_disease/svc_trained_model.pkl", "rb"))
print("Model loaded from heart_disease/svc_trained_model.pkl")
````

### Step 7.3: Evaluate the Machine Learning Model

#### Step 7.3.1: Make Predictions with the Trained Models on Testing Data

> **Understanding:** The model predicts the 60 unseen patients. The first 15 predictions are shown next to the true labels; all 60 are saved to `model-predictions.csv`.

````python
"""
Purpose of this section:
-------------------------
Uses the trained model to predict the testing data.
"""

model_predictions = model.predict(input_vector_test)

pd.options.mode.chained_assignment = None
testing_data_encoded["Predictions"] = model_predictions
testing_data_encoded.to_csv(r"heart_disease/model-predictions.csv", index=False, header=True)

model_predictions = testing_data_encoded
print("\n\nPredictions Returned by svc_trained_model (first 15 of", len(model_predictions), "):")
print("=====================================================================\n")
print(model_predictions.head(15))
````

### Step 7.4: Calculate the Accuracy Score

> **Understanding:** Accuracy = correct predictions / 60 testing patients.

````python
"""
Purpose of this section:
-------------------------
Calculates the accuracy of the model on the testing data.
"""

model_accuracy_score = accuracy_score(
    model_predictions["HeartDisease"],
    model_predictions["Predictions"]
)

print("\n\nAccuracy Score:")
print("===============\n")
print(round(model_accuracy_score, 2))
print("Correct predictions:", (model_predictions["HeartDisease"] == model_predictions["Predictions"]).sum(),
      "out of", len(model_predictions))
````

### Step 7.5: Detailed Evaluation Metrics (Precision, Recall, F1, Confusion Matrix)

> **Understanding:** In medicine, **Recall for the "Yes" class** is especially important: it shows how many patients who really have heart disease the model catches. A missed patient (false negative) is more dangerous than a false alarm.

````python
"""
Purpose of this section:
-------------------------
Prints precision, recall, F1-score and the confusion matrix.
"""

print("\nClassification Report:")
print("=======================\n")
print(classification_report(
    model_predictions["HeartDisease"],
    model_predictions["Predictions"],
    target_names=["No Heart Disease (0)", "Heart Disease (1)"]
))

conf_matrix = confusion_matrix(model_predictions["HeartDisease"], model_predictions["Predictions"])
conf_matrix_df = pd.DataFrame(
    conf_matrix,
    index=["Actual: No", "Actual: Yes"],
    columns=["Predicted: No", "Predicted: Yes"]
)
print("\nConfusion Matrix:")
print("==================\n")
print(conf_matrix_df)

plt.figure(figsize=(5, 4))
sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues",
            xticklabels=["No", "Yes"], yticklabels=["No", "Yes"])
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - SVC on Heart Disease Testing Data")
plt.tight_layout()
plt.show()
````

### Step 10.3: The Web App (`streamlit-app/app.py`)

> **Understanding:** The app (1) loads each `.pkl` model, (2) recreates the LabelEncoders, (3) shows drop-down menus for the 4 inputs, and (4) calls `model.predict()` when the button is clicked. Its `requirements.txt` pins the same library versions used to train the models, so the saved models load correctly on the cloud.

```python
"""
AIC354 Assignment 2: End-to-End Machine Learning System
Zain Aftab (FA24-BSE-155), COMSATS University Islamabad, Lahore Campus
-----------------------------------------------------------------------
One Streamlit web app that serves both trained models:
  Tab 1: Titanic Passenger Survival Prediction (Support Vector Classifier)
  Tab 2: Heart Disease Prediction (Support Vector Classifier)

Each tab loads the .pkl model saved by the notebook and recreates the
same LabelEncoders (fitted on the same fixed category lists), so the
predictions match the notebook exactly.
"""

import pickle
from pathlib import Path

import pandas as pd
import streamlit as st
from sklearn.preprocessing import LabelEncoder

st.set_page_config(page_title="AIC354 ML Prediction Systems", page_icon="🩺", layout="centered")

APP_DIR = Path(__file__).parent


@st.cache_resource
def load_model(file_name):
    with open(APP_DIR / file_name, "rb") as f:
        return pickle.load(f)


# Same category lists used to train the LabelEncoders in the notebook
TITANIC_CATEGORIES = {
    "PClass": ["First", "Second", "Third"],
    "Gender": ["Male", "Female"],
    "Sibling": ["Zero", "One", "Two", "Three"],
    "Embarked": ["Southampton", "Cherbourg", "Queenstown"],
}

HEART_CATEGORIES = {
    "ChestPain": ["Typical Angina", "Atypical Angina", "Non-Anginal Pain", "Asymptomatic"],
    "Slope": ["Upsloping", "Flat", "Downsloping"],
    "MajorVessels": ["Zero", "One", "Two", "Three"],
    "Thal": ["Normal", "Fixed Defect", "Reversible Defect"],
}


@st.cache_resource
def build_encoders(categories):
    encoders = {}
    for column, values in categories.items():
        encoders[column] = LabelEncoder().fit(values)
    return encoders


def encode(user_input, encoders):
    encoded = user_input.copy()
    for column in user_input.columns:
        encoded[column] = encoders[column].transform(user_input[column])
    return encoded


st.title("Machine Learning Prediction Systems")
st.markdown(
    "**AIC354 Machine Learning Fundamentals, Assignment 2**  \n"
    "Zain Aftab (FA24-BSE-155), COMSATS University Islamabad, Lahore Campus"
)

titanic_tab, heart_tab = st.tabs(["🚢 Titanic Survival", "🩺 Heart Disease"])

# ---------------------------------------------------------------------------
# Tab 1: Titanic Passenger Survival Prediction
# ---------------------------------------------------------------------------
with titanic_tab:
    st.subheader("Titanic Passenger Survival Prediction")
    st.write("Pick a passenger's details. The trained Support Vector Classifier predicts whether they survived.")

    col1, col2 = st.columns(2)
    with col1:
        pclass = st.selectbox("Passenger Class (PClass)", TITANIC_CATEGORIES["PClass"], index=2)
        gender = st.selectbox("Gender", TITANIC_CATEGORIES["Gender"])
    with col2:
        sibling = st.selectbox("Siblings/Spouses Aboard", TITANIC_CATEGORIES["Sibling"])
        embarked = st.selectbox("Port of Embarkation", TITANIC_CATEGORIES["Embarked"])

    if st.button("Predict Survival", type="primary", use_container_width=True):
        user_input = pd.DataFrame({
            "PClass": [pclass], "Gender": [gender], "Sibling": [sibling], "Embarked": [embarked],
        })
        encoded = encode(user_input, build_encoders(TITANIC_CATEGORIES))
        prediction = load_model("titanic_svc_trained_model.pkl").predict(encoded)[0]
        if prediction == 1:
            st.success("### ✅ SURVIVED")
        else:
            st.error("### ❌ NOT SURVIVED")
        with st.expander("Encoded feature vector sent to the model"):
            st.dataframe(encoded, hide_index=True)

# ---------------------------------------------------------------------------
# Tab 2: Heart Disease Prediction
# ---------------------------------------------------------------------------
with heart_tab:
    st.subheader("Heart Disease Prediction")
    st.write("Enter a patient's details. The trained Support Vector Classifier predicts whether heart disease is present.")

    col1, col2 = st.columns(2)
    with col1:
        chest_pain = st.selectbox("Chest Pain Type", HEART_CATEGORIES["ChestPain"], index=3)
        slope = st.selectbox("Slope of Peak Exercise ST Segment", HEART_CATEGORIES["Slope"], index=1)
    with col2:
        major_vessels = st.selectbox("Major Vessels Coloured by Fluoroscopy", HEART_CATEGORIES["MajorVessels"])
        thal = st.selectbox("Thal (Thallium Stress Test)", HEART_CATEGORIES["Thal"])

    if st.button("Predict Heart Disease", type="primary", use_container_width=True):
        user_input = pd.DataFrame({
            "ChestPain": [chest_pain], "Slope": [slope],
            "MajorVessels": [major_vessels], "Thal": [thal],
        })
        encoded = encode(user_input, build_encoders(HEART_CATEGORIES))
        prediction = load_model("heart_svc_trained_model.pkl").predict(encoded)[0]
        if prediction == 1:
            st.error("### ⚠️ HEART DISEASE PRESENT")
        else:
            st.success("### ✅ NO HEART DISEASE")
        with st.expander("Encoded feature vector sent to the model"):
            st.dataframe(encoded, hide_index=True)

st.divider()
st.caption(
    "Built for teaching purposes. Titanic: 100-row simplified dataset. "
    "Heart Disease: UCI Cleveland dataset (297 complete records, 4 input attributes). "
    "Not a medical diagnosis tool."
)
```
